## 1️⃣ What Is a Selection Structure?

A selection structure allows a program to **choose what to do** based on a condition.

Think about everyday decisions:

- **If** it is raining, bring an umbrella.
- **If** your score is 75 or higher, you pass.
- **If** the traffic light is red, stop.

A computer uses the same basic idea:

**Condition → True or False → Choose an action**

## 2️⃣ Conditions: True or False?

A **condition** is something that can be evaluated as either `True` or `False`.

Examples:

- `10 > 5` → `True`
- `10 < 5` → `False`
- `75 >= 75` → `True`
- `20 == 10` → `False`

Run the cell below and observe the results.

````python
print(10 > 5)
print(10 < 5)
print(75 >= 75)
print(20 == 10)
````

## 3️⃣ Comparison Operators

| Operator | Meaning | Example |
|---|---|---|
| `==` | equal to | `5 == 5` |
| `!=` | not equal to | `5 != 3` |
| `>` | greater than | `8 > 5` |
| `<` | less than | `3 < 5` |
| `>=` | greater than or equal to | `75 >= 75` |
| `<=` | less than or equal to | `70 <= 75` |

⚠️ Remember: `=` means **assign a value**, while `==` means **compare two values**.

## 4️⃣ The `if` Statement


## 5️⃣ The `if-else` Statement

What if we want the program to do something when the condition is **False** too?

Use `if-else`:

```python
if condition:
    action_if_true
else:
    action_if_false
```

Example: A score of 75 or higher is `Passed`; otherwise, it is `Failed`.

````python
score = 68

if score >= 75:
    print("Passed")
else:
    print("Failed")
````

## 6️⃣ Getting Input from the User

Programs can make decisions using information entered by the user.

For numbers, we commonly use `int(input(...))`.

Example:

```python
age = int(input("Enter your age: "))
```

````python
age = int(input("Enter your age: "))

if age >= 12:
    print("You are 12 or older.")
else:
    print("You are younger than 12.")
````

## 7️⃣ Multiple Decision Paths: `if-elif-else`

Sometimes there are **more than two possible outcomes**.

Python uses `elif` (short for *else if*) to check another condition.

Example grading system:

- 90 and above → A
- 80–89 → B
- 70–79 → C
- Below 70 → D

````python
score = int(input("Enter your score: "))

if score >= 90:
    print("Grade: A")
elif score >= 80:
    print("Grade: B")
elif score >= 70:
    print("Grade: C")
else:
    print("Grade: D")
````

## 8️⃣ Boundary Conditions

A **boundary condition** checks whether an input is within an allowed range.

Example from the study guide: a valid score is from **0 to 100**.

We can check both limits using `and`:

```python
if score >= 0 and score <= 100:
    print("Valid Score")
else:
    print("Invalid Score")
```

````python
score = int(input("Enter a score: "))

if score >= 0 and score <= 100:
    print("Valid Score")
else:
    print("Invalid Score")
````

## 9️⃣ Boolean Logic: `and` and `or`

Selection structures can combine conditions.

### `and`
Both conditions must be true.

Example:
```python
age >= 12 and age <= 18
```

### `or`
At least one condition must be true.

Example:
```python
student == "Y" or age >= 65
```

These ideas are useful when a decision depends on more than one piece of information.
