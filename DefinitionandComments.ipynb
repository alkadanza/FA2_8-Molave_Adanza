```python
import math

# Get a coordinate from the user
def get_coordinate(name):
    return float(input(f"Enter {name}: "))

# Calculate the distance between two points
def calculate_distance(x1, y1, x2, y2):
    return math.sqrt((x2 - x1)**2 + (y2 - y1)**2)

# Main program
def main():
    # Get the coordinates
    x2 = get_coordinate("x2")
    x1 = get_coordinate("x1")
    y2 = get_coordinate("y2")
    y1 = get_coordinate("y1")

    # Calculate the distance
    distance = calculate_distance(x1, y1, x2, y2)

    # Display the result
    print("The distance between the two points is:", distance)

main()

#Reflection: while doing this activity the math library help me simplify the program by making it show and not complicated,the functions on how to solve the equation was made easier by the math library. If the program were to be made without the pow() and sqrt() it will be much harder because the code will be more complicated.
```
